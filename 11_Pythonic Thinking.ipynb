```python
#enumerate : returns the (index,item) tuple
```

```python
characters = ["Krillin","Goku", "Vegeta", "Gohan", "Piccolo"]

list(enumerate(characters))
```

```python
for index, character in enumerate(characters):
    print(index,character)
```

```python
characters = ["Krillin","Goku", "Goku", "Gohan", "Piccolo",
              "Krillin","Goku", "Vegeta", "Gohan", "Piccolo",
              "Piccolo","Goku", "Vegeta", "Goku", "Piccolo"]

character_map={character:[] for character in characters}
print(character_map)
```

```python
for index, character in enumerate(characters):
    character_map[character].append(index)
print(character_map)
```

```python
#zip : combine iterables into tupels
```

```python

latitude = [4,5,6,4,6,2,4,5,6]
longitude = [6,3,6,3,4,5,6,8,5]

[*zip(latitude,longitude)]
```

```python
altitude = [12,41,15,16,15,23,14,51,61]

[*zip(latitude, longitude, altitude)]
```

```python
short = [1,2,3,4]
long = [1,2,3,4,5,6,7,8]

[*zip(short, long)]
```

```python
from itertools import zip_longest
```

```python
short = [1,2,3,4]
long = [1,2,3,4,5,6,7,8]

[*zip_longest(short, long, fillvalue=None)]
```

```python
#LBYL(look before you leap) : check first if the file exists
```

```python
import os
LBYL='LBYL.txt'
if os.path.isfile(LBYL):
    with open(LBYL,'a') as f:
        f.write('This file exists!')
else:
    print('This file dosent exist')
```

```python
EAFP='EAFP.txt'
try:
    with open(EAFP,'a') as f:
        f.write('This file exists!')
except Exception as e:
    print(f'This file dosent exist, because {e}')
```
