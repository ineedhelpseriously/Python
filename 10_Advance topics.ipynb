```python
#Basic Unpacking
a,b,c=[1,2,3]
print(f'a: {a},b: {b},c: {c}')

#Extended Iterable Unpacking with *
a,b,*c=[1,2,3,4,5]
print(f'a: {a},b: {b},c: {c}')

a,*b,c=[1,2,3,4,5]
print(f'a: {a},b: {b},c: {c}')

#Ignoring Values

a, _, c=[1,2,3]
```

```python
#Unpacking Nested data
data= ("ineedhelpseriously",(18,"Student"))
name,(age,profession)= data
```

```python
#Unpacking in Functional Arguments
def print_names(*names):
    for name in names:
        print(name)
        
print_names("A","B","C")
```

```python
#Combining and Unpacking
list1=[1,2,3]
list2=[4,5,6]
combined=[*list1,*list2]
print(combined)
print(*list1)
```

```python
dict1={"a":1,"b":2}
dict2={"c":3,"d":4}
combined_dict={**dict1, **dict2}
print(combined_dict)
```

```python
#Swapping using unpacking
x=1
y=2
print(f"{x},{y}")
x,y=y,x
print(f"{x},{y}")
```

```python
def demo_exec():
    code="""def greet(name):
    return f"Hello, {name}\""""
    
    local_scope={}
    exec(code,{},local_scope)
    print(local_scope["greet"]("ineedhelpseriously"))
    
demo_exec()
```

```python
def demo_eval():
    expression=input("Type an exp: ")
    result=eval(expression)
    print(f"{result}")

demo_eval()
```

```python
def SafeDemoEval():
    expression=input("Type an expression that uses a,b,c")
    variable={"a":2,"b":3,"c":4}
    result=eval(expression,{},variable)
    print(f"{result}")

SafeDemoEval()
```

```python
cube=[_**3 for _ in range(10) if _%2 == 0]
print(cube)
```

```python
names: list[str]=['alice','bob','cattor']
long_names: list[str]=[]

for _ in names:
    if len(_)>4:
        long_names.append(_)
print(long_names)
```

```python
names: list[str]=['alice','bob','cattor']
long_names: list[str]=[_ for _ in names if len(_)>4]
print(long_names)
print([_ for _ in names if len(_)>4])
```

```python
class MyRange:
    def __init__(self,start,end):
        self.value=start
        self.end=end
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.value>=self.end:
            raise StopIteration
        current= self.value
        self.value+=1
        return current

def my_range(start,end):
    current=start
    while current<end:
        yield current
        current+=1
        

```

```python
nums=MyRange(1,10)
```

```python
print(next(nums))
```

```python
for _ in nums:
    print(_)
```

```python
num=my_range(1,10)
for _ in num:
    print(_)
```
