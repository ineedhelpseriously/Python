```python
f=open('test.txt','r')
print(f.name)
print(f.mode)
f.close()
```

```python
#context manager
with open('test.txt', 'r') as f:
    pass
print(f.closed)
```

```python
with open('test.txt','r') as f:
    f_contents=f.read()
    print(f_contents)
```

```python
with open('test.txt','r') as f:
    f_contents_lines=f.readlines()
    print(f_contents_lines)
```

```python
with open('test.txt','r') as f:
    f_contents_line=f.readline()
    print(f_contents_line)
    f_contents_line=f.readline()
    print(f_contents_line)
    f_contents_line=f.readline()
    print(f_contents_line)
```

```python
with open('test.txt', 'r') as f:
    for line in f:
        print(line, end='')
```
